# **Анализ пользовательского поведения и коммерческой эффективности на e-commerce платформе**

Проект по анализу данных из реального датасета.  

---

## Описание датасета

Логи событий пользователей на e-commerce платформе:
- Просмотры, добавления в корзину, покупки
- ~400K строк, 10 столбцов
- Период: ~2 месяца
---

## Очистка данных

Датасет проверен на:
- Несоответствие типов данных
- Наличие пропусков значений
- Наличие дубликатов

И очищен от найденных артефактов
